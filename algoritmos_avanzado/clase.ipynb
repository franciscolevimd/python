```python
def busqueda_binaria(arr, objetivo):
    bajo = 0
    alto = len(arr) -1

    while bajo <= alto:
        medio = (bajo + alto) // 2
        if arr[medio] == objetivo:
            return medio
        elif arr[medio] < objetivo:
            bajo = medio + 1
        else:
            alto = medio - 1
            
    return -1

'''
Notación Big O:
O(log n)

Expresión Matemática:

Tamaño:
0 - n
1 - n/2
2 - n/4
3 - n/8

n/2^k donde k es el número de iteraciones.

Peor caso es tamaño 1 (todo)

n/2^k = 1 (2^k) ambos lados
n= 2^k log base 2 ambos lados
log(2)n = k
'''
```

## <font color='green'>Quicksort</font>

**Quicksort** es un aloritmo de ordenammiento eficiente basado en la técnica de **divide y vencerás**. Es conocido por su rapidez en la mayoría de los casos y su simplicidad en términos de implementación.

En el caso promedio. O(nlogn)
En el peor caso: O(n²) ocurre cuando el pivote seleccionado es el menor o mayor elemento repetidamente, por ejemplo, si el arreglo ya está ordenado o casi ordenado.

- <font color='green'>Procesos:</font> Se selecciona un elemento del arreglo, llamado pivote.

- <font color='green'>Particionar:</font> Se reorganiza el arreglo de manera que todos los elementos menores que le pivote queden antes que él y todos los mayores queden después.

- <font color='green'>Ordenera sub arreglos:</font> Se aplica recursivamente el mismo proceso a los subarreglos que resultan de la partición.

- <font color='green'>Combinar sub arreglos:</font> El arreglo está ordenado cuando todos los subarreglos han sido ordenados.

```python
def quicksort(arr):
    # Caso base: una lista de 0 o 1 elementos está ordenada
    if len(arr) <= 1:
        return arr
    else:
        # Elegir el pivote (en este caso el elemento centrla de la lista)
        pivot =  arr[len(arr) // 2]
        # Particionar la lista en tres sublistas
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        rigtht = [x for x in arr if x > pivot]
        # ordenar sublistas y concatenarlas
        print(f"pivot: {pivot}, quicksort: {left}, middle: {middle}, quicksort: {rigtht}")
        return quicksort(left) + middle + quicksort(rigtht)
    
    
# Ejemplo de uso
lista = [3, 6, 8, 10, 1, 2, 1]
print(f"Lista ordenada: {quicksort(lista)}")
```

```python
def quicksort_inplace(arr, low, high):
    if low < high:
        # Particiona la lista y obtiene el ínidice del pivote.
        pi = partition(arr, low, high)
        # Ordena las sublistas.
        quicksort_inplace(arr, low, pi - 1)
        quicksort_inplace(arr, pi + 1, high)
        
def partition(arr, low, high):
    # Elige el pivote (último elemento).
    pivot = arr[high]
    # Índice del elemento más pequeño.
    i = low - 1
    for j in range(low, high):
        if arr[j] < pivot:
            i += 1
            # Intercambio de elementos.
            arr[i], arr[j] = arr[j], arr[i]
    # Mueve el pivote a su lugar.
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

            
```

## <font color='green'>Dijkstra</font>

El algoritmo de **Dijkstra** encuentra el camino más corto desde un nodo fuente a todos los otros nodos en un grafo ponderado con pesos de aristas no negativos.

Complejidad O((V+E)logV)

- <font color='green'>Inicialización:</font> Distancia del nodo fuente a él mismo como 0. Marca las distancias de los demás nodos como infinita. Marca los nodos como no visitados. Inicia una cola de prioridad con el nodo fuente, donde se guardarán los nodos con menor distancia conocidos.

- <font color='green'>Proceso:</font> Mientras haya nodos no visitados, extrae el nodo con la distancia más corta y actualiza las distancias hacia sus vecinos si encuentra un camino más corto. Marca el nodo como visitado.

- <font color='green'>Repetición:</font> Se repite el proceso hast que los nodos han sido visitados.



```python
import heapq

def dijkstra(graph, start):
    # Inicialización.
    distances = {node: float('inf') for node in graph}
    distances[start] = 0
    priority_queue = [(0, start)]  # (distancia, nodo)
    visited = set()

    while priority_queue:
        current_distance, current_node = heapq.heappop(priority_queue)
        print(f"Current_distance: {current_distance}, current_node: {current_node}")
    
        if current_node in visited:
            continue
        visited.add(current_node)
        # Actualización de las distancias a los nodos vecinos.
    
        for neighbor, weight in graph[current_node].items():
            print(f"Neighbor: {neighbor}, weight: {weight}")
            distance = current_distance + weight
            print(f"Distance: {distance}, distance neighbor: {distances[neighbor]}")
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                heapq.heappush(priority_queue, (distance, neighbor))                
        print(f"queue: {priority_queue}, distances: {distances}")
        
    return distances


# Ejemplo de uso.
graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'A': 1, 'C': 2, 'D': 5},
    'C': {'A': 4, 'B': 2, 'D': 1},
    'D': {'B': 5, 'C': 1}
}

start_node = 'A'
distances = dijkstra(graph, start_node)
print(f"Distancias desde el nodo {start_node}: {distances}")
```

```python
import networkx as nx

# Crea grafo dirigido con pesos
G = nx.DiGraph()
G.add_weighted_edges_from([
    ('A', 'B', 1),
    ('A', 'C', 4),
    ('B', 'C', 2),
    ('B', 'D', 5),
    ('C', 'D', 1)
])
 
# Calcular el camino más corto desde nodo 'A'.
distancias = nx.single_source_dijkstra_path_length(G, 'A')
print(f"Distancias desde el nodo A: {distancias}") 
```

## <font color='green'>BST</font>

Un **Árbol de búsqueda binaria (BST, por sus siglas en inglés)** es una estructura de datos fundamental en la informática que se utiliza para almacenar datos de manera que se puedan buscar, insertar y eliminar de manera eficiente.

**En el peor caso (árbol desbalanceado):** O(n), donde n es el número de nodos.
**En el mejor caso (árbol balanceado):** O(logn)

- <font color='green'>Propiedad del BST:</font> Para cualquier nodo en el árbol, el valor de todos los nodos en el subárbol izquierdo es menor que el valor del nodo. Cada nodo tiene un valor (clave), un hijo izquierdo y un hijo derecho.

- <font color='green'>Búsqueda:</font> Encontrar un nodo con un valor específico.

- <font color='green'>Inserción:</font> Agregar un nuevo nodo al árbol manteniendo la propiedad del BST.

- <font color='green'>Eliminación:</font> Eliminar un nodo del árbol y reorganizar el árbol para mantener la propiedad del BST.

- <font color='green'>Recorridos:</font> Visitar todos los nodos del árbol en un orden específico (inorden, preorden, postorden).

```python
class Node:
    def __init__(self, key):
        self.left = None
        self.right = None
        self.value = key
        
class BST:
    def __init__(self):
        self.__root = None
        
    def search(self, key):
        return self._search(self.__root, key)
    
    def _search(self, node, key):
        if node is None or node.value == key:
            return node
        
        if key < node.value:
            return self._search(node.left, key)
        
        return self._search(node.right, key)
```

## <font color='green'>K-means</font>

El algoritmo K-means es un método de clustering no supervisado que se utiliza para agrupar datos en K clusters (grupos) basados en características similares.

Complejidad O(n⋅k⋅d⋅i), donde n es el número de puntos de datos, k es el número de clusters, d es la dimensionalidad de los datos, e i es el número de iteraciones hasta la convergencia.

- <font color='green'>Inicialización:</font> Selecciona KKK puntos al azar del conjunto de datos como los centroides iniciales de los clústeres.

- <font color='green'>Asignación:</font> Asigna cada punto de datos al cluster cuyo centroide esté más cercano (basado en la distancia Euclidiana, por ejemplo).

- <font color='green'>Actualización:</font> Calcula los nuevos centroides como el promedio de todos los puntos asignados a cada clúster.

- <font color='green'>Iteración:</font> Repite los pasos de asignación y actualización hasta que los centroides ya no cambien significativamente o se alcance un número máximo de iteraciones.

```python
# Importar las bibliotecas necesarias
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
# Generar datos de ejemplo
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# Crear y ajustar el modelo K-means
kmeans = KMeans(n_clusters=4)
kmeans.fit(X)

# Obtener los centroides y las etiquetas de los clústeres
centroids = kmeans.cluster_centers_
labels = kmeans.labels_

# Visualizar los resultados
plt.scatter(X[:, 0], X[:, 1], c=labels, s=50, cmap='viridis')
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', s=200, alpha=0.75, marker='X')  # Centroides
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('K-means Clustering')
plt.show()
```
