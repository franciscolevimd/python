# <font color='green'>Magic Methods</font>

Los métodos mágicos o dunder functions son un conjunto de métodos predefinidos en Python que nos proveen funciones sintácticas especiales. Estos nos permiten que los objetos propios se comporten similar a los tipos incorporados de Python. Se pueden reconocer fácilmente porque inician y terminan con dos guiones bajos, por ejemplo `__init__`, `__call__` y `__len__`

`__init__`

Nos permite inicializar un objeto nuevo.

```python
class Pizza:
    def __init__(self, size, toppings):
        self.size = size
        self.toppings = toppings


# Now let's create a pizza
my_pizza = Pizza("large", ["pepperoni", "mushrooms"])        

print(my_pizza.size)  # This will print. large
print(my_pizza.toppings)  # This will print: ['pepperoni', 'mushrooms']
```

```python
print(my_pizza)
```

`__str__`

Nos permite definir la descripición o representación de strings de nuestro objeto.

```python
class Pizza:
    def __init__(self, size, toppings):
        self.size = size
        self.toppings = toppings
        
    def __str__(self):
        return f"A {self.size} pizza with {', '.join(self.toppings)}"


# Now let's create a pizza
my_pizza = Pizza("large", ["pepperoni", "mushrooms"])        

print(my_pizza.size)  # This will print. large
print(my_pizza)  # This will print: ['pepperoni', 'mushrooms']
```

`__repr__`

Nos permite crear una representación formal y detallada de la descripción del objeto. Si `__str__` no es definida para un objeto, automáticamente se toma `__repr__` para mostrar el objeto o convertirlo en cadena.

```python
class Pizza:
    def __init__(self, size, toppings):
        self.size = size
        self.toppings = toppings
        
    def __repr__(self):
        return f"Pizza {self.size}, {self.toppings}"


# Now let's create a pizza
my_pizza = Pizza("large", ["pepperoni", "mushrooms"])

print(my_pizza.size)  # This will print. large
print(my_pizza)  # This will print: ['pepperoni', 'mushrooms']
```

`__add__`

Nos permite definir el comportamiento de nuestro objeto cuand llamamos al operador de suma `+`.

```python
class Pizza:
    def __init__(self, size, toppings):
        self.size = size
        self.toppings = toppings
        
    def __add__(self, other):
        if not isinstance(other, Pizza):
            raise TypeError("You can only add another Pizza!")
        new_toppings = self.toppings + other.toppings
        return Pizza(self.size, new_toppings)


# Let's create two pizzas
pizza1 = Pizza('large', ["pepperoni", "mushrooms"])
pizza2 = Pizza('medium', ["olives", "pineapple"])

# And now let's "add" them
combined_pizza = pizza1 + pizza2

print(combined_pizza.toppings)  # This will print: 'pepperoni', 'mushrooms', 'olives', 'pineapple'
print(combined_pizza.size)
```

`__len__`

Nos permite definir que retorna cuando llamamos a la función len() con nuestro objeto.

```python
class Pizza:
    def __init__(self, size, toppings):
        self.size = size
        self.toppings = toppings
        
    def __len__(self):
        return len(self.toppings)


# Now let's create a pizza
my_pizza = Pizza("large", ["pepperoni", "mushrooms"])

len(my_pizza)
```

`__iter__`

Perite que nuestro objeto sea un iterable, osea, puede ser usado en un ciclo `for`. Para hacer esto debemos definir también `__next__`, esta es usada para definir el valor que retorna en la siguiente iteración.

```python
class Pizza:
    def __init__(self, size, toppings):
        self.size = size
        self.toppings = toppings
        
    def __iter__(self):
        self.n = 0
        return self
    
    def __next__(self):
        if self.n < len(self.toppings):
            result = self.toppings[self.n]
            self.n += 1
            return result
        else:
            raise StopIteration
        
# Let's create a pizza
my_pizza = Pizza("large", ["pepperoni", "mushrooms", "olives"])

# And now let's iterate over it
for topping in my_pizza:
    print(topping)
```

# <font color='green'>Properties</font>

Las properties son la forma pythonica de evitar la creación de métodos para obtener y modificar atributos de una clase. Esta función nos ayuda a convertir atributos de una clase en **properties** o **managed attribute**.

## <font color='green'>Creando properties con la inicialización de la función property()</font>

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    def _get_radius(self):
        """The radius property"""
        print("Get radius")
        return self._radius

    def _set_radius(self, value):
        print("Set radius")
        self._radius = value

    def _del_radius(self):
        print("Delete radius")
        del self._radius
        
    radius = property (
        fget=_get_radius,
        fset=_set_radius,
        fdel=_del_radius,
        doc="The radius property."
    )
```

```python
circle = Circle(42.0)
```

```python
circle.radius
```

```python
circle.radius = 50
```

```python
circle.radius
```

```python
del circle.radius
```

```python
circle.radius
```

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        """The radius property"""
        print("Get radius")
        return self._radius

    @radius.setter
    def radius(self, value):
        print("Set radius")
        self._radius = value

    @radius.deleter
    def radius(self):
        print("Delete radius")
        del self._radius
```

```python
circle = Circle(42.0)
```

```python
circle.radius
```

```python
circle.radius = 23
```

```python
circle.radius
```

```python
del circle.radius
```

# <font color='green'>Métodos de instancia, clase y estáticos</font>

Los <font color='green'>métodos de instancia</font> son creados para modificar un objeto instanciado de una clase.

Los <font color='green'>métodos de clase</font> trabajan drectamente con la clase, desde que su parámetro es la clase en sí.

Los <font color='green'>métodos estátios</font> no saben nada acerca de la clase, solo trabajan con los parámetros recibidos.

```python
class MyClass:
    # Método de instancia.
    def method(self):
        return f"Método de instancia llamado, {self}"
    
    # Método de clase.
    @classmethod
    def class_method(cls):
        return f"Método de clase llamado, {cls}"
    
    # Método de estátco.
    @staticmethod
    def static_method():
        return "Método de estático llamado"
```

```python
obj = MyClass()
```

```python
obj.method()
```

```python
MyClass.method(obj)
```

```python
obj.class_method()
```

```python
obj.static_method()
```

```python
MyClass.class_method()
```

```python
MyClass.static_method()
```

```python
MyClass.method()
```

# <font color='green'>Dataclasses</font>

Las dataclasses son una nueva forma de definir clases introducida en Python 3.7. Estos vienen con funcionalidades básicas que ya están implementadas ahorrandonos líneas de código.

```python
from dataclasses import dataclass

@dataclass
class Position:
    name: str
    lon: float = 0.0
    lat: float = 0.0
```

## <font color='green'>Clases inmutables</font>

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class Position:
    name: str
    lon: float = 0.0
    lat: float = 0.0
```

```python
pos = Position('Oslo', 10.8, 59.9)
```

```python
pos.name
```

```python
pos.name = 'México'
```
