# <font color='green'>Itertools</font>

Este módulo implementa un número de piezas básicas iterator inspiradas en cosntucts de APL, haskell y SML. Cada pieza ha sido reconvertida a una froma apropiada para Python.

El módulo estandariza un conjunto base de herramientas rápidas y eficientes en memoria, útiles pro sí mismas o en combinación con otras. Juntas, forma un <<ágebra de iteradores>>, haciendo posible las contraucción de herramientas especializadas, sucintas y eficientes, en Python puro.

## <font color='green'>itertools.compress</font>

Crea un iterador qeu devuelve elementos de los datos donde el elemento correspondiente en los selectores es verdadero.

Se detiene cuando se han agotado los iterables de los datos o de los selectores.

```python
from itertools import compress

nombres = ['Janeth', 'Francisco', 'Briana', 'Jessica', 'Juan', 'Ulises']
filtro = [0, 1, 0, 0, 1, 1]

compress_iter = compress(nombres, filtro)
print(list(compress_iter))
```

## <font color='green'>itertools.count</font>

Crea un iterador que devuelva valores espaciados de manera uniforme, comenzando con start.

Se puede utilizar con `map()` para generar puntos de datos consecutivos o con `zip()`, para agregar números de secuencia.

```python
from itertools import count

# Funciona como un range pero de forma infinita.
count_iter = count(start=1, step=2)

print(next(count_iter))
print(next(count_iter))
print(next(count_iter))

print("Inicio del for.")
for i in count_iter:
    print(i)
    if i > 10:
        break
print("Fin del for.")

print(next(count_iter))
print(next(count_iter))
print(next(count_iter))
```

## <font color='green'>itertools.cycle</font>

Crea un iterador que devuelve elementos del iterable y gaurda una copia de cada uno. Cuando se agota el iterable devuelve elementos de la copia guardada. Se repite indefinidamente.

```python
from itertools import cycle

numeros = [7, 0, 1]
numeros_cycle = cycle(numeros)

print(next(numeros_cycle))
print(next(numeros_cycle))
print(next(numeros_cycle))
print(next(numeros_cycle))
print(next(numeros_cycle))
print(next(numeros_cycle))
print(next(numeros_cycle))
```

## <font color='green'>itertools.permutations</font>

Devuelve permutaciones sucesivas de longitud r de elementos del iterable.

```python
from itertools import permutations

nombres = ["Briana", "Jessica", "Zyanya"]

print(f"Permutaciones de {nombres}")
permutaciones_iter = permutations(nombres)
for index, dato in enumerate(permutaciones_iter, start=1):
    print(f"Permutación {index}, {dato}")
```

## <font color='green'>itertools.combinations</font>

Retorna subsecuencias de longitud r con elementos del iterable de entrada.

```python
from itertools import combinations

nombres = ["Diana", "Rebeca", "Rocío"]

print(f"Combinaciones de {nombres}")
combinations_iter = combinations(nombres, 2)
for index, comb in enumerate(combinations_iter):
    print(f"Combinación {index}: {comb}")
```

## <font color='green'>itertools.combinations_with_replacement</font>

Retorna subsecuencias de longitud r con elementos del iterable de entrada, permitiendo que haya elementos individuales repetidos más de una vez.

```python
from itertools import combinations_with_replacement

nombres = ["Diana", "Rebeca", "Rocío"]

print(f"Combinaciones con elementos repetidos de {nombres}")
combinations_iter = combinations_with_replacement(nombres, 2)
for index, comb in enumerate(combinations_iter):
    print(f"Combinación con reemplazo: {index}: {comb}")
```

## <font color='green'>itertools.chain, intertools.chain.from_iterable</font>

Crea un iterador que retorna elementos del primer iterable hasta que es consumido, para ñuego proceder con el siguiente iterable, hasta que todos los iterables son consumidos.

Se utiliza para tratar secuencias consecutivas como una sola suencia.

```python
from itertools import chain

nombres = ["Parkesita", "Mónica", "María"]
numeros = "123456"

print(list(chain(nombres, numeros)))
```

# <font color='green'>Functools</font>

El módulo `functools` es para funciones de orden superior. En general, cualquier objeto invocable puede ser tratado como una funcción para los propósitos de este módulo.

## <font color='green'>functools.cache</font>

Caché de funciones limitado, ligero y simple. A avces llamado <<memorización>>.

```python
from functools import cache

@cache
def factorial(n):
    print(f"Llamado de la función con valor {n}")
    if n == 1:
        print(f"Resultado de la función con valor {n} = {1}")
        return 1

    x = n * factorial(n - 1)
    print(f"Resultado de la funcción con valor {n} = {x}")
    return x

print(f"Cache inicial: {factorial.cache_info()}")
print(f"Factorial 10: {factorial(10)}")
print(f"Cache 10: {factorial.cache_info()}")
print("----------------")
print(f"Factorial 12: {factorial(12)}")
print(f"Cache 12: {factorial.cache_info()}")
print("----------------")
print(f"Factorial 15: {factorial(15)}")
print(f"Cache 15: {factorial.cache_info()}")
print("----------------")
print(f"Factorial 6: {factorial(6)}")
```


## <font color='green'>functools.partial</font>

Retorna un nuevo partial object que cuando sea llamado se comportará como `func` llamado con los argumentos pasados previamente.


```python
from functools import partial

def aplicar_descuento(precio, descuento):
    return precio - (precio * descuento)

# Versiones especialzadas de la función de descuentos.
descuento_10 = partial(aplicar_descuento, descuento=0.10)
descuento_20 = partial(aplicar_descuento, descuento=0.20)
descuento_30 = partial(aplicar_descuento, descuento=0.30)

# Lista de precios.
precios = [100, 200, 300, 400, 500]
print(f"Preciones originales: {precios}")

# Aplicar diferentes descuentos a los precios
print(f"Precios con 10% de descuento: {list(map(descuento_10, precios))}")
print(f"Precios con 20% de descuento: {list(map(descuento_20, precios))}")
print(f"Precios con 30% de descuento: {list(map(descuento_30, precios))}")
```

## <font color='green'>functools.reduce</font>

Retorna un nuevo partial object que cuando sea llamado se comportará como `func` llamado con los argumentos pasados previamente.

```python
from functools import reduce

def suma(acumulador, valor):
    resultado = acumulador + valor
    print(f"{acumulador = }, {valor = }, {resultado = }")
    return resultado

datos = list(range(1, 8))
print(f"Datos: {datos}")
resultado = reduce(suma, datos, 10)
print(resultado)
```

## <font color='green'>functools.singledispatch</font>

Transforma una función en una single-dispatch generic function.

```python
from functools import singledispatch

@singledispatch
def suma(a, b):
    raise NotImplementedError("Unsuportes type")

@suma.register(int)
def _(a, b):
    print(f"Type's argument: {type(a)}")
    return a + b

@suma.register(list)
def _(a, b):
    print(f"Type's argument: {type(a)}")
    return sum(a) + sum(b)

print(suma([3, 4, 5], [6, 5, 7]))
print(suma(1, 10))
```
