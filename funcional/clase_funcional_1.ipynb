# <font color='green'>Programación Funcional.</font>

La programación funcional es un paradigma de programación declarativo donde el foco está en las <font color='green' size='4'>funciones matemáticas puras</font> y el uso de <font color='green' size='4'>expresiones en lugar de setencias</font> que alteren el estado del programa.

Las funciones son tratadas como <font color='green' size='4'>ciudadanos de primera clase</font>, lo que significa que pueden ser asignadas a variables, pasadas como argumentos y retornadas como un resultado.

### <font color='green'>Paradigmas de programación.</font>

Un paradigma de programación es un enfoque o conjunto de principios que guían la forma en que se estructura y se escribe el código en un lenguaje de programación.

Define la forma en que se modelan y resuelven los problemas utilizando dicho lenguaje.

### <font color='green'>Programación Imperativa vs Declarativa.</font>

- **Imperativa.** Describe <font color='green' size='4'>qué pasos</font> hay que dar para obtener la solución de un problema.
- **Declarativa.** Se enfoca en describir <font color='green' size='4'>qué/cuál</font> es la soluciòn sin entrar en los detalles de su control de flujo.

## <font color='green'>Caraterísticas.</font>

- <font color='green' size='4'>**Inmutabilidad:**</font> No se modifican variables ni se cambia el estado. En su lugar, se crean nuevos valores.
- <font color='green' size='4'>**Funciones puras:**</font> Las funciones no tienen efectos secundarios. El mismo conjunto de entradas produce la misma salida siempre.
- <font color='green' size='4'>**Declarativo:**</font> El codigodescribe el qué hacer en lugar del cómo hacerlo.
- <font color='green' size='4'>**Control impícito:**</font> Se usan expresiones como `map()`, `filter()`, `reduce()` para aplicar operaciones sn necesidad de bulces explícitos.

### <font color='green'>Funciones puras.</font>

- Devuelven el mismo resultado al ser llamada con los mismos argumentos de entrada.
- El resultado solo depende de la entrada: no tienen memoria.
- No tiene efectos colaterales (side effects).

### <font color='green'>Funciones como ciudadanos de primer orden.</font>

Se refieren a elementos en un lenguaje de programación que pueden ser tratados de manera completa y sin restriccciones.

Estos elementos pueden ser asignados a variables, alamacenados en estructurasd de datos, pasados como argumentos a funciones y devueltos como resultados de funciones.

### <font color='green'>Funciones de orden superior.</font>

Las funciones que operan con otras funciones se conocen como funciones de orden superior en Python: si la función contiene otras funciones como parámetro de entrada o si devuelve una función como salida.

### <font color='green'>Funciones recursivas.</font>

Una función recursiva es una función que se define en términos de si misma a tarvés de una expresión autorreferencial. Eso significa que la función se llama a sí misma y repite el comportamiento hasta que se cumple alguna condición, para finalmente devolver un resultado.

### <font color='green'>Currying.</font>

El currying es una técnica de programación funcional en la que una función con múltiples argumentos se transforma en una <font color='green' size='4'>secuencia de funciones</font>, cada una con uun solo argumento.

En otras palabras, una función que toma varios parámetros se convierte en una serie de funciones anidadas, donde cada función toma n solo arguento y devuelve una nueva función.

## <font color='green'>Lambdas, Maps and Filters.</font>

### <font color='green'>Lambdas.</font>

Son funciones anónimas que solo pueden contener una expresión. `lambda <parameter_list>: <expresion>`

```python
def get_even(x):
    return x%2 == 0

print(get_even)

print(lambda x: x%2 == 0)
```

### <font color='green'>Función Map().</font>

La función `map()` devuelve un objeto `Map` (que es un iterador) con los resultados después de aplicar la función dada a cada elemento de un iterable dado (lista, tupla, etc).

Sintaxis: `map(<f>, <iterable>)`

```python
animals = ["cat", "dog", "bird", "gecko"]
iterator = map(lambda animal: animal[::-1], animals)

for i in iterator:
    print(i)
```

```python
"+".join(map(str, [1, 2, 3, 4, 5]))
```

```python
hello_world = 'Hola Garlopaz'
iterator = map(lambda x: x.upper(), hello_world)
print(''.join(iterator))
print(hello_world)
```

#### <font color='green'>Llamando `map()` con múltiples iterables.</font>

Sintaxis: `map(<f>, <iterable_1>, <iterable_2>, ... <iterable_n>)`

```python
def f(a, b, c):
    return a + b + c

resultado = list(map(f, [1, 2, 3], [10, 20, 30], [100, 200, 300]))
print(resultado)
```

### <font color='green'>Función Filter().</font>

La función `filter()` filtra la secuencia dada con ayuda de una función que prueba si cada elemento de la secuencia es verdadero o no.

Sintaxis: `filter(<f>, <iterable>)`

```python
list(filter(lambda x: x > 100, [1, 111, 2, 222, 3, 333]))
```
