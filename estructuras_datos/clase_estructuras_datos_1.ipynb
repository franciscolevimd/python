# <font color='green'>Pilas</font>

Una pila es una estrucutra de datos que sigue el principio de **último en entrar, primero en salir (LIFO)**. Esto significa que el último elemento que se añade a la pila es el primero en ser eliminado. Puedes pensar en una pila como una pila de platos: el plato que pones en la parte superor es el primero que sacarás.

```python
from collections import deque

class PilaDeque:
    
    def __init__(self):
        self.__items = deque()
```

### <font color='green'>Push</font>

Añade un elemento a la parte superior de la pila.

```python
from collections import deque

class PilaDeque:
    
    def __init__(self):
        self.__items = deque()

    def agregar(self, item):
        self.__items.append(item)
```

### <font color='green'>Pop</font>

Elimina el último elemento que se agregó a la pila y lo devuelve.

```python
from collections import deque

class PilaDeque:
    
    def __init__(self):
        self.__items = deque()

    def eliminar(self):
        if not self.esta_vacia():
            return self.__items.pop()
        raise IndexError("Eliminar de una pila vacía")
```

### <font color='green'>Peek</font>

Devuelve el último elemento agregado sin eliminarlo.

```python
from collections import deque

class PilaDeque:
    
    def __init__(self):
        self.__items = deque()

    def cima(self):
        if not self.esta_vacia():
            return self.__items[-1]
        raise IndexError("Pila vacía")
```

### <font color='green'>Is Empty</font>

Devuelve True si la pila está vacía.

```python
from collections import deque

class PilaDeque:
    
    def __init__(self):
        self.__items = deque()

    def esta_vacia(self):
        return len(self.__items) == 0
```

### <font color='green'>Size</font>

Devuelve el tamaño de la pila.

```python
from collections import deque

class PilaDeque:
    
    def __init__(self):
        self.__items = deque()

    def tamano(self):
        return len(self.__items)
```

# <font color='green'>Colas</font>

Una cola es una estructura de datos que sigue el principio de **primero en entrar, primero en salir (FIFO)**. Imagina una fila en una tienda: la primera persona en entrar es la primera en ser atendida.

```python
from collections import deque

class ColaDeque:
    
    def __init__(self):
        self.__items = deque()
```

### <font color='green'>Enqueue</font>

Añadir un elemento al final de la cola.

```python
from collections import deque

class ColaDeque:
    
    def __init__(self):
        self.__items = deque()
        
    def encolar(self, item):
        self.__items.append(item)
```

### <font color='green'>Denqueue</font>

Retirar el elemento del frente de la cola.

```python
from collections import deque

class ColaDeque:
    
    def __init__(self):
        self.__items = deque()

    def desencolar(self):
        if not self.esta_vacia():
            return self.__items.popleft()
        raise IndexError("Desencolar de una cola vacía")
```

### <font color='green'>Peek</font>

Obtener el elemento en el frente de la cola sin eliminarlo.

```python
from collections import deque

class ColaDeque:
    
    def __init__(self):
        self.__items = deque()

    def frente(self):
        if not self.esta_vacia():
            return self.__items[0]
        raise IndexError("Cola vacía")
```
