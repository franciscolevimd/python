```python
from threading import Thread


count = 0


def deposit():
    global count
    for x in range(1_000_000):
        count += 1


def withdraw():
    global count
    for x in range(1_000_000):
        count -= 1

t1 = Thread(target=deposit)
t2 = Thread(target=withdraw)


t1.start()
t2.start()

t1.join()
t2.join()

print(count)
```

```python
from threading import Thread, Lock


lock = Lock()
count = 0


def deposit():
    global count
    for x in range(1_000_000):
        lock.acquire()
        count += 1
        lock.release()


def withdraw():
    global count
    for x in range(1_000_000):
        lock.acquire()
        count -= 1
        lock.release()

t1 = Thread(target=deposit)
t2 = Thread(target=withdraw)


t1.start()
t2.start()

t1.join()
t2.join()

print(count)
```

```python
from threading import Thread, Lock
from threading import Lock
from threading import Event
from itertools import cycle


lock = Lock()
event = Event()
count = 0


def waiting_for():
    print("Comenzamos los depósitos y balances...")
    for c in cycle(r"-\\|/-"):
        print(c, end="\r", flush=True)
        if event.wait(0.1):
            break


def deposit():
    global count
    for x in range(10_000_000):
        lock.acquire()
        count += 1
        lock.release()


def withdraw():
    global count
    for x in range(10_000_000):
        lock.acquire()
        count -= 1
        lock.release()

t1 = Thread(target=deposit)
t2 = Thread(target=withdraw)
t3 = Thread(target=waiting_for)


t1.start()
t2.start()
t3.start()
t1.join()
t2.join()

event.set()
t3.join()
```

```python
import time
from threading import Thread


def fibonnaci(number):
    if number == 1:
        return 0
    
    if number == 2:
        return 1
    
    return fibonnaci(number - 1) + fibonnaci(number -2)


start = time.time()
numbers = [38, 39, 40, 41, 42]
threads = []

for number in numbers:
    t = Thread(target=lambda: print( f"El resultado de {number} es {fibonnaci(number)}" ))
    t.start()
    threads.append(t)

for t in threads:
    t.join()
    
print(f"Tiempo tomado para resolver Fibonnaci: {time.time() - start} segundos.")

```

```python
from threading import Thread
import time

def fibonnaci(number):
    if number == 1:
        return 0
    
    if number == 2:
        return 1
    
    return fibonnaci(number - 1) + fibonnaci(number - 2)


def show_fibonnaci_info(number):
    print(fibonnaci(number))


start = time.time()

thread1 = Thread(target=show_fibonnaci_info, args=(42,))
thread2 = Thread(target=show_fibonnaci_info, args=(42,))

thread1.start()
thread2.start()

thread1.join()
thread2.join()

print(f"Fin de programa {time.time() - start}")
```
