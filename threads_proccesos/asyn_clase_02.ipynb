# <font color='green'>¿Qué es una pool?.</font>

- Es una estructura que permite gestionar la creación y la ejecución de
threads y/o procesos.

- Permite crear una cantidad **Limitada** de Threads y/o procesos.

- Permite re-utilizar Threads y/o procesos que se encuentren
disponibles, pudiendo así mejorar el performance de la aplicación.

## <font color='green'>Creación de una pool</font>

`from concurrent.futures import ThreadPoolExecutor`

`with ThreadPoolExecutor() as executor: ...`



# <font color='green'>Futuros</font>

## <font color='green'>¿Qué es un futuro?</font>

Es una abstracción que permite llevar la gestión de un resultado
asíncrono. Algo que aún no sucede, pero, **posiblemente**, suceda.

```python
import time
from concurrent.futures import Future


def hello_world(future):
    print("Oli")

future = Future()
future.add_done_callback(hello_world)

time.sleep(3)

future.set_result(True)
```

```python
import time
from concurrent.futures import Future


def callback1(future):
    print("Oli")

def callback2(future):
    print("Oli")

def callback3(future):
    print("Oli")


future = Future()
future.add_done_callback(callback1)
future.add_done_callback(callback2)
future.add_done_callback(callback3)

time.sleep(3)

future.set_result("Hola mundo")
```

```python
import time
from concurrent.futures import Future


def callback1(future):
    result = future.result()
    print(result)

def callback2(future):
    print("Oli")

def callback3(future):
    print("Oli")


future = Future()
future.add_done_callback(callback1)
future.add_done_callback(callback2)
future.add_done_callback(callback3)

time.sleep(3)

future.set_result("Hola mundo")
```

# <font color='green'>Coroutines</font>

## <font color='green'>¿Qué son las Coroutines?</font>

- Es una función que pueda pausar su ejecución y reanudarla en otro
momento.

- La función se reanudará cuando la tarea haya sido completada.


```python
import time
from random import randint


# Generadores
def random_numbers_gen():
    for _ in range(10):
        print("Antes de pausar")
        yield randint(10, 100)
        print("Después de pausar")

for x in random_numbers_gen():
    print(x)
```

# <font color='green'>¿Threads, Procesos o Async?</font>

- La corutinas permite ejecutar tareas de manera concurrente dentro de
un único hilo/proceso.

- Los hilos permiten la ejecución concurrente de varias tareas dentro
de un único proceso.

- Los procesos son unidades de ejecución completamente
independientes. Se aprovechan todos los recursos de la
computadora.
